```python
# Assignment 1, Question 1
def concrete_cost():
    cement_cost = float(input("Please enter the cost of cement: ："))
    sand_cost = float(input("Please enter the cost of sand："))
    volume = float(input("Please enter the volume of cement: ："))


    total_parts = 2 + 2.5
    cement_volume = volume * 2 / total_parts
    sand_volume = volume * 2.5 / total_parts

    total_cost = cement_volume * cement_cost + sand_volume * sand_cost
    print(f" the cost of manufacturing the premix concrete：{total_cost:.2f}")


concrete_cost()
```

```python
# Assignment 1, Question 2
total_income = float(input("please enter the total income: "))
total_deduction = float(input("please enter the total deduction: "))
total_allowance = float(input("please enter the total allowance: "))

Net_ChargeableIncome = total_income - total_deduction - total_allowance
Net_Income = total_income - total_deduction

if Net_ChargeableIncome <= 0:
    Personal_Income_Tax = 0
elif Net_ChargeableIncome <= 200000:
    Personal_Income_Tax = Net_ChargeableIncome * 0.05
elif Net_ChargeableIncome <= 400000:
    Personal_Income_Tax = ((Net_ChargeableIncome - 200000) * 0.09) + 10000
elif Net_ChargeableIncome <= 600000:
    Personal_Income_Tax = ((Net_ChargeableIncome - 400000) * 0.13) + 28000
elif Net_ChargeableIncome <= 800000:
    Personal_Income_Tax = ((Net_ChargeableIncome - 600000) * 0.17) + 54,000
else:
    Personal_Income_Tax = ((Net_ChargeableIncome - 800000) * 0.2) + 88,000

upper_limit_tax = Net_ChargeableIncome * 0.17
tax_final = min(upper_limit_tax, Personal_Income_Tax)
print(f"The amount of personal income tax is: {tax_final:.2f}")

```

```python
def main():
    # months is used as a lookup table
    months = "JanFebMarAprMayJunJulAugSepOctNovDec"
    n = int(input("Enter a month number (1-12): "))
    # compute starting position of month n in months
    pos = (n-1) * 3

    # Grab the appropriate slice from months
    monthAbbrev = months[pos:pos+3]
    # print the result
    print ("The month abbreviation is", monthAbbrev + ".")
main()
```

```python
def main():
    print ("This program creates a file of usernames from a")
    print ("file of names.")
    infileName = input("What file are the names in? ")
    outfileName = input("What file should the usernames go in? ")
    infile = open(infileName, 'r')
    outfile = open(outfileName, 'w')
    for line in infile:
        first, last = line.split()
        uname = (first[0]+last[:7]).lower()
        print(uname, file=outfile)
    infile.close()
    outfile.close()
    print("Usernames have been written to", outfileName)
main()
```

```python
# def is_anagram(a,b):
#     a = a.lower()
#     b = b.lower()
#     return sorted(a) == sorted(b)
#
# print(is_anagram("abc","bac"))

def is_anagram(a, b):
    a, b = a.lower(), b.lower()
    if len(a) != len(b):
        return False

    count = {}  # 统计 s1 的字母频次
    for char in a:
        # 呼应录音：首次出现默认为0，再+1
        count[char] = count.get(char, 0) + 1
    print(count)

    for char in b:
        if char not in count or count[char] == 0:
            return False
        count[char] -= 1
    print("Two strings are anagrams")
    return True



is_anagram("abc","bac")
```

```python
def is_rotational(s1, s2):

    if len(s1) != len(s2):
        return False

    return s2 in (s1 + s1)

s1 = input()
s2 = input()
print(is_rotational(s1, s2))
```

```python
N = 2000
list = [1] * N        # 创建 2000 个元素，全部初始化为 1

list[0] = 0           # 0 和 1 不是素数，预先划掉
list[1] = 0

for x in range(2, N):        # 从 list[2] 开始
    if list[x] == 1:          # 只有还活着的元素，才去划它的倍数
        for multiple in range(x * 2, N, x):   # x 的倍数，且大于 x
            list[multiple] = 0

# 输出所有值为 1 的下标
for i in range(2, N):
    if list[i] == 1:
        print(i)
```
